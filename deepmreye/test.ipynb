```python
from deepmreye.preprocess import load_label 
```

```python
import os
base_path = "/data2/2104/derivatives/"
calibration_labels = os.path.join(base_path, "deepmreye/")
```

```python
import numpy as np
this_label = load_label(label_path=calibration_labels, label_type="calibration_run")

# Reshape to show timepoints × channels
reshaped_label = this_label.reshape(this_label.shape[0], -1)

# Create column headers
headers = []
for i in range(this_label.shape[1]):  # for each channel
    headers.extend([f'Channel_{i}_X', f'Channel_{i}_Y'])

header_str = ','.join(headers)
np.savetxt('this_label_structured.csv', reshaped_label, delimiter=',', header=header_str, comments='')
```
