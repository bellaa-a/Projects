````python
import heapq, math, copy
````

````python
def dist(current, goal):
    """
    Calculates the Euclidean distance between two points in a 2D grid.

    Args:
        current (Tuple[int, int]): The current position (x, y).
        goal (Tuple[int, int]): The target position (x, y).

    Returns:
        float: The Euclidean distance between the current position and the goal.
    """
    x, y = goal[0] - current[0], goal[1] - current[1]
    return math.sqrt(x**2 + y**2)
````

````python
def evaluate_move(next_position, cost_tracker, current_position, previous_positions, search_queue, grid, start, goal):
    """
    Evaluates and updates the movement of a position during the A* search algorithm.

    Checks if the move is valid (not out of bounds or blocked), calculates the cost for the move,
    and if it improves the path to the goal, updates the relevant cost trackers and adds the move to the priority queue.

    Args:
        next_position (Tuple[int, int]): The position to evaluate (row, col).
        cost_tracker (dict): A dictionary that tracks the cost to reach each position.
        current_position (Tuple[int, int]): The current position (row, col).
        previous_positions (dict): A dictionary that stores the previous position of each cell.
        search_queue (list): A priority queue (min-heap) for the A* search.
        grid (List[List[str]]): The 2D grid representing the map, where obstacles are marked with '⬛'.
        start (Tuple[int, int]): The starting position (row, col).
        goal (Tuple[int, int]): The goal position (row, col).

    Returns:
        None: This function updates the cost tracker, previous positions, and search queue in-place.
    """
    if (0 <= next_position[0] < len(grid) and 0 <= next_position[1] < len(grid[0]) and
        grid[next_position[0]][next_position[1]] != '⬛' and next_position != start):
        new_cost = cost_tracker[current_position] + 1
        estimated_cost = new_cost + dist(next_position, goal)
        if next_position not in cost_tracker or estimated_cost < cost_tracker[next_position]:
            cost_tracker[next_position] = new_cost
            previous_positions[next_position] = current_position
            heapq.heappush(search_queue, (estimated_cost, next_position))
````

````python
def find_path(grid, start, goal, directions):
    """
    Finds the optimal path from the start position to the goal using the A* search algorithm.

    This function uses a priority queue (min-heap) to explore the grid, evaluates possible moves
    using the A* algorithm, and constructs the shortest path from the start to the goal by following
    the previous positions.

    Args:
        grid (List[List[str]]): The 2D grid representing the world, where obstacles are marked with '⬛' and free spaces with '⬜'.
        start (Tuple[int, int]): The starting position (row, col).
        goal (Tuple[int, int]): The goal position (row, col).
        directions (List[Tuple[int, int]]): List of possible movement directions (e.g., up, down, left, right).

    Returns:
        List[Tuple[int, int]]: A list of positions representing the optimal path from start to goal, inclusive.
                               Returns None if no path is found.
    """
    search_queue = [(0, start)]
    previous_positions = {start: None}
    cost_tracker = {start: 0}
    path = []
    while search_queue:
        _, current_position = heapq.heappop(search_queue)
        if current_position == goal:
            while current_position:
                path.append(current_position)
                current_position = previous_positions[current_position]
            path.reverse()
            return path
        for direction in directions:
            next_position = (current_position[0] + direction[0], current_position[1] + direction[1])
            evaluate_move(next_position, cost_tracker, current_position, previous_positions, search_queue, grid, start, goal)
    return None
````

````python
def add_repulsion(repulsion, entity, other_entity):
    """
    Updates the repulsion vector based on the distance between two entities. 
    The function calculates the repulsive force applied to an entity when it's too close to another entity, 
    and also adds extra repulsion if they overlap.

    Args:
        repulsion (List[float]): The current repulsion vector, modified by this function.
        entity (Tuple[int, int]): The position (row, col) of the first entity.
        other_entity (Tuple[int, int]): The position (row, col) of the second entity.

    Returns:
        List[float]: The updated repulsion vector after considering the distance and overlap between the two entities.
    """
    distance = (other_entity[0] - entity[0], other_entity[1] - entity[1])
    norm = math.sqrt(distance[0]**2 + distance[1]**2)
    if norm < 2 and norm > 0: repulsion[0] -= distance[0] / (norm ** 2); repulsion[1] -= distance[1] / (norm ** 2)
    if norm < 1: 
        overlap = 1 - norm
        repulsion[0] += (distance[0] / norm) * overlap * 10
        repulsion[1] += (distance[1] / norm) * overlap * 10
    return repulsion
````

````python
def add_cohesion(cohesion, entity, neighbors, cohesion_strength=0.05, cohesion_radius=5):
    """
    Adds a weak attraction force towards nearby butterflies to encourage flocking behavior.

    Args:
        cohesion (List[float]): The current cohesion vector.
        entity (Tuple[int, int]): The position of the current butterfly.
        neighbors (List[Tuple[int, int]]): The positions of nearby butterflies.
        cohesion_strength (float): Scaling factor for cohesion force.
        cohesion_radius (int): Maximum distance for cohesion to take effect.

    Returns:
        List[float]: The updated cohesion vector.
    """
    center_of_mass = [0.0, 0.0]
    count = 0

    for neighbor in neighbors:
        if neighbor != entity:
            dist_x, dist_y = neighbor[0] - entity[0], neighbor[1] - entity[1]
            distance = math.sqrt(dist_x**2 + dist_y**2)
            if 0 < distance < cohesion_radius:  # Only apply cohesion within the defined radius
                center_of_mass[0] += neighbor[0]
                center_of_mass[1] += neighbor[1]
                count += 1

    if count > 0:
        center_of_mass[0] /= count
        center_of_mass[1] /= count
        cohesion[0] += (center_of_mass[0] - entity[0]) * cohesion_strength
        cohesion[1] += (center_of_mass[1] - entity[1]) * cohesion_strength

    return cohesion
````

````python
def calculate_forces(entity, robot_position, swarm):
    """
    Calculates the attraction and repulsion forces on an entity from the robot and other butterflies in the swarm.

    - **Attraction**: Towards the robot.
    - **Repulsion**: From the robot and nearby butterflies.
    - **Cohesion**: Weak attraction towards other butterflies to maintain flocking behavior.

    Args:
        entity (Tuple[int, int]): The position of the butterfly.
        robot_position (Tuple[int, int]): The position of the robot.
        swarm (List[Tuple[int, int]]): List of butterfly positions.

    Returns:
        Tuple[List[float], List[float], List[float]]: Attraction, repulsion, and cohesion force vectors.
    """
    attraction = [robot_position[0] - entity[0], robot_position[1] - entity[1]]
    norm_attraction = math.sqrt(attraction[0]**2 + attraction[1]**2)
    attraction = [attraction[0] / norm_attraction, attraction[1] / norm_attraction] if norm_attraction > 0 else attraction

    repulsion = [0.0, 0.0]
    cohesion = [0.0, 0.0]

    for other_entity in swarm + [robot_position]:
        if other_entity != entity:
            repulsion = add_repulsion(repulsion, entity, other_entity)

    cohesion = add_cohesion(cohesion, entity, swarm)

    return attraction, repulsion, cohesion
````

````python
def update_swarm_positions(swarm, robot_position, grid):
    """
    Updates butterfly positions based on attraction to the robot, repulsion from others, and cohesion.

    Args:
        swarm (List[Tuple[int, int]]): List of butterfly positions.
        robot_position (Tuple[int, int]): Robot's position.
        grid (List[List[str]]): 2D grid of the environment.

    Returns:
        List[Tuple[int, int]]: Updated butterfly positions.
    """
    updated_positions = []
    for entity in swarm:
        attraction, repulsion, cohesion = calculate_forces(entity, robot_position, swarm)
        new_position = (
            entity[0] + attraction[0] + repulsion[0] + cohesion[0],
            entity[1] + attraction[1] + repulsion[1] + cohesion[1]
        )
        new_position = (
            min(max(new_position[0], 0), len(grid) - 1),
            min(max(new_position[1], 0), len(grid[0]) - 1)
        )
        updated_positions.append(new_position)
    return updated_positions
````

````python
def run_simulation():
    """
    Runs the simulation of the robot navigating the grid towards the goal while avoiding obstacles
    and interacting with the swarm of entities. It updates the positions of the robot and swarm after
    each move and displays the current state of the grid.

    This function assumes the existence of:
        - `find_path` to compute the robot's path.
        - `update_swarm_positions` to update the positions of swarm entities.
    """
    temp_world = copy.deepcopy(world)
    swarm = [(start[0], start[1]+2), (start[0], start[1]+4), (start[0], start[1]+6), (start[0], start[1]+8)]
    path = find_path(temp_world, start, goal, moves)
    
    if path:
        first_iteration = True
        for robot_position in path:
            grid_copy = [row[:] for row in temp_world]
            grid_copy[robot_position[0]][robot_position[1]] = '🤖'
            grid_copy[goal[0]][goal[1]] = '❤️'
            for entity in swarm:
                grid_copy[int(entity[0])][int(entity[1])] = '🦋'
            print("\n".join("".join(row) for row in grid_copy), "\n")
            swarm = update_swarm_positions(swarm, robot_position, temp_world)
            if first_iteration:
                temp_world[start[0]][start[1]] = '⬜'
                temp_world[20][2] = '⬜'
                temp_world[20][4] = '⬜'
                temp_world[20][6] = '⬜'
                temp_world[20][8] = '⬜'
                first_iteration = False
````

# Problem #2: Game Theory

We are going to conduct an experiment where two agents (players) employ the *Best Response with Inertia* algorithm against each other.

## Best Response with Inertia Algorithm

Here is the outline of the algorithm for each player:


```
On day zero, play a random action

On day t, do one of two things:

    * With high probability, play the best response to what your opponent did yesterday, or
    * With low probability, be lazy and just do what you did yesterday.
```

## Game

The particular 2x2 game being played is the Stag-Hunt game, with the matrix:


|  | | |
|---|---|---|
|  | S | H |
| S | 3, 3 | 0, 2 |
| H | 2, 0 | 2, 2 |


Player 1 (P1) is the `row` player; Player 2 (P2) is the `col` player.

````python
import random
import matplotlib.pyplot as plt
````

## get_utility

This function takes a joint action tuple and returns the utility for each player, packaged as a tuple. It is hardcoded for the Stag-Hunt game described above.

* **a** Tuple: joint action. a could be ('S', 'H'), for instance, which would signify that P1 has chosen to do 'S' and P2 has chosen to do 'H'.


**returns** 

* **utility** Tuple: the utility for each player as (utility of P1, utility of P2). For example, if the input is ('S', 'H'), the returned value is (0, 2) -- 0 for P1, 2 for P2. 

````python
def get_utility(a):
    if a[0] == 'S' and a[1] == 'S':
        return (3, 3)
    elif a[0] == 'S' and a[1] == 'H':
        return (0, 2)
    elif a[0] == 'H' and a[1] == 'S':
        return (2, 0)
    elif a[0] == 'H' and a[1] == 'H':
        return (2, 2)
````

````python
def run_experiment(num_trials, num_days, debug=True):
    """
    Runs an experiment simulating multiple trials where two players (P1 and P2) take actions over multiple days.
    The experiment tracks the actions and utilities for each player and prints results if debugging is enabled.
    
    In this experiment, the players' actions are based on inertia. Each player has a 75% chance of repeating the 
    previous action based on the opponent's last action, or they will repeat their own previous action.

    Args:
        num_trials (int): The number of trials to run in the experiment. Each trial simulates a sequence of days.
        num_days (int): The number of days each trial lasts. Each day the players will choose actions.
        debug (bool): If True, prints the trial and day number along with actions and utilities of both players.

    Returns:
        None: This function does not return any value, but it modifies the global histories and prints results if debugging is enabled.

    Global Variables:
        actions_P1 (list): A list of possible actions for player 1.
        actions_P2 (list): A list of possible actions for player 2.
        history_P1_action (list): A list storing the history of player 1's actions.
        history_P2_action (list): A list storing the history of player 2's actions.
        history_P1_utility (list): A list storing the history of player 1's utility after each action.
        history_P2_utility (list): A list storing the history of player 2's utility after each action.
        get_utility (function): The function that calculates the utilities for the players based on their actions.
    """
    if debug: print(f"Trial\tDay\tP1\tP2\n--------------------------")
    for trial in range(num_trials):  
        if debug: print(f"--------------------------")
        action_P1 = random.choice(actions_P1)
        action_P2 = random.choice(actions_P2)
        for day in range(num_days):
            if day > 0:
                if random.random() < 0.75:
                    action_P1 = 'S' if history_P2_action[-1] == 'S' else 'H'
                else:
                    action_P1 = history_P1_action[-1]
                if random.random() < 0.75:
                    action_P2 = 'S' if history_P1_action[-1] == 'S' else 'H'
                else:
                    action_P2 = history_P2_action[-1]
            # Don't modify below:
            utility_P1, utility_P2 = get_utility(tuple((action_P1, action_P2)))
            history_P1_action.append(action_P1)
            history_P2_action.append(action_P2)
            history_P1_utility.append(utility_P1)
            history_P2_utility.append(utility_P2)
            if debug: print(f"{trial}\t{day}\t{action_P1}\t{action_P2}")
````

````python
# Parameters. Don't modify. Not even debug flag. 
run_experiment(num_trials=20, num_days=10, debug=True)
````

````python
def plot_results(history_P1_action, history_P2_action):
    labels = ['(S,S)', '(S,H)', '(H,S)', '(H,H)']
    cell_counts = [0, 0, 0, 0]
    for i in range(len(history_P1_action)):
        if history_P1_action[i] == 'S' and history_P2_action[i] == 'S':
            cell_counts[0]+=1
        if history_P1_action[i] == 'S' and history_P2_action[i] == 'H':
            cell_counts[1]+=1
        if history_P1_action[i] == 'H' and history_P2_action[i] == 'S':
            cell_counts[2]+=1
        if history_P1_action[i] == 'H' and history_P2_action[i] == 'H':
            cell_counts[3]+=1
    plt.bar(labels, cell_counts)
    plt.ylim(0, 200)
    plt.show()
````

````python
# Visualize the results. Don't modify. 
plot_results(history_P1_action, history_P2_action)
````
