## Game

* We have two players: our original good robot and the new bad robot '👾'
* At each turn of the game, a player gets to move in the world
* Players only get to move according to `moves`
* '⬛' are obstacles that neither robot can traverse
* Each player knows the other player's current location
* Each player knows the location of the '❤️'
* '🤖' wants to get to the '❤️' without being zapped
* '👾' is trying to zap '🤖'
* '🤖' wins if it gets to the '❤️'
* '👾' wins if it zaps the good robot
* The game is over when either player wins or we've reached 82 plies (41 good robot moves and 41 bad robot moves)

## Adversarial Search


The MINIMAX pseudocode is as follows:

```
MINIMAX(state) returns action
    return argmax action in Actions MIN-VALUE(RESULT(state,action))
    

MAX-VALUE(state) returns value
if TERMINAL(state) return UTILITY(state)
v <- -inf
for each action in Actions(state)
    v <- MAX(v, MIN-VALUE(RESULT(state, action))
return v


MIN-VALUE(state) returns value
if TERMINAL(state) return UTILITY(state)
v <- +inf
for each action in Actions(state)
    v <- MIN(v, MAX-VALUE(RESULT(state, action))
return v
```


## Notes. 

* MINIMAX is used to model the opponent as a rational player that will also try it's best to win. When moves are simulated in the tree, that's just what they are, simulations. The actual board only changes once MINIMAX returns an action and that action is taken.

````python
# Your code starts here.
import copy
import random
import math
````

````python
def euclidean_dist(pos1, pos2):
    """
    Calculates the Euclidean distance between two points in a 2D grid.

    Args:
        pos1 (Tuple[int, int]): The first position (row, col).
        pos2 (Tuple[int, int]): The second position (row, col).

    Returns:
        float: The Euclidean distance between the two points.
    """
    return math.sqrt((pos1[0] - pos2[0]) ** 2 + (pos1[1] - pos2[1]) ** 2)
````

````python
# euclidean_dist unit tests
assert euclidean_dist((3, 3), (3, 3)) == 0.0   # Same point
assert euclidean_dist((2, 1), (2, 4)) == 3.0   # Horizontal
assert euclidean_dist((1, 5), (4, 5)) == 3.0   # Vertical
assert euclidean_dist((0, 0), (3, 4)) == 5.0   # Diagonal (3-4-5 triangle)
assert abs(euclidean_dist((1, 1), (4, 5)) - 5.0) < 1e-6  # Arbitrary points
assert abs(euclidean_dist((1000, 2000), (3000, 6000)) - 4472.135955) < 1e-6  # Large numbers

print("All tests passed!")
````

````python
def good_utility(good_pos, bad_pos, goal):
    """
    Computes the utility for the good robot.

    The utility function prioritizes reaching the goal while avoiding the bad robot.
    - If the good robot can be zapped next turn (within √2 distance), it gets the lowest possible score (-∞).
    - Otherwise, the utility is the difference between its distance from the bad robot and its distance to the goal.

    Args:
        good_pos (Tuple[int, int]): The position of the good robot (row, col).
        bad_pos (Tuple[int, int]): The position of the bad robot (row, col).
        goal (Tuple[int, int]): The position of the goal (row, col).

    Returns:
        float: A utility score where higher values are better for the good robot.
    """
    if euclidean_dist(good_pos, bad_pos) <= math.sqrt(2):  # If good can be zapped next turn
        return float('-inf')
    return euclidean_dist(good_pos, bad_pos) - euclidean_dist(good_pos, goal)
````

````python
# good_utility unit tests
assert good_utility((3, 3), (3, 3), (5, 5)) == float('-inf')  # Zapped immediately
assert good_utility((2, 2), (3, 3), (5, 5)) == float('-inf')  # Zapped next turn (diagonal)
assert good_utility((1, 1), (4, 4), (5, 5)) < 0  # Closer to goal than to bad (fixed)
assert good_utility((5, 5), (1, 1), (6, 6)) > 0  # Safe and near goal
assert good_utility((5, 5), (1, 1), (0, 0)) < 0  # Moving away from goal
assert good_utility((3, 3), (0, 0), (1, 1)) > 0  # Closer to bad than goal

print("All tests passed!")
````

````python
def bad_utility(bad_pos, good_pos, goal, question):
    """
    Computes the utility for the bad robot.

    The bad robot wants to minimize the distance to the good robot (to zap it)
    and, in some cases, also consider the goal position.

    - If `question == 2`: The utility is negative distance to the good robot (prioritizing chasing the good robot).
    - If `question == 3` or `question == 4`: The utility considers both the distance to the good robot 
      and the distance to the goal, ensuring a balance between pursuit and positioning.

    Args:
        bad_pos (Tuple[int, int]): The position of the bad robot (row, col).
        good_pos (Tuple[int, int]): The position of the good robot (row, col).
        goal (Tuple[int, int]): The position of the goal (row, col).
        question (int): Determines the utility function variant.

    Returns:
        float: A utility score where lower values indicate a better position for the bad robot.
    """
    if question == 2:
        return - euclidean_dist(bad_pos, good_pos)
    elif question == 3 or question == 4:
        return - euclidean_dist(bad_pos, good_pos) - euclidean_dist(bad_pos, goal)   
````

````python
# bad_utility unit tests
assert bad_utility((3, 3), (5, 5), (7, 7), 2) == -euclidean_dist((3, 3), (5, 5))  # Only chasing 🤖
assert bad_utility((3, 3), (5, 5), (7, 7), 3) == -euclidean_dist((3, 3), (5, 5)) - euclidean_dist((3, 3), (7, 7))  # Chasing and goal
assert bad_utility((3, 3), (5, 5), (7, 7), 4) == -euclidean_dist((3, 3), (5, 5)) - euclidean_dist((3, 3), (7, 7))  # Same as question 3
assert bad_utility((0, 0), (1, 1), (2, 2), 2) < 0  # Should be negative (closer is better)
assert bad_utility((6, 6), (2, 2), (0, 0), 3) < bad_utility((6, 6), (4, 4), (0, 0), 3)  # Being closer to 🤖 & ❤️ should be worse (higher negative value)

print("All tests passed!")
````

````python
def minimax(position, opp_position, goal, depth, maximizing_player, world, is_good, question):
    """
    Implements the Minimax algorithm for decision-making in a two-player adversarial game.

    The algorithm searches for the best move for the current player by recursively evaluating possible future positions.
    The good robot ('🤖') maximizes its utility, while the bad robot ('👾') minimizes the good robot’s utility.

    Args:
        position (Tuple[int, int]): The current position of the player making a move.
        opp_position (Tuple[int, int]): The opponent’s position.
        goal (Tuple[int, int]): The goal position.
        depth (int): The remaining depth of search.
        maximizing_player (bool): True if the current player is maximizing (good robot), False otherwise.
        world (List[List[str]]): The 2D grid representing the game world.
        is_good (bool): True if the player using this function is the good robot, False otherwise.
        question (int): The specific evaluation strategy for the bad robot.

    Returns:
        Tuple[float, Tuple[int, int]]: The best evaluation score and the corresponding move.
    """
    if depth == 0:
        return (good_utility(position, opp_position, goal) if is_good 
                else bad_utility(position, opp_position, goal, question)), position

    best_eval, best_move = (float('-inf'), None) if maximizing_player else (float('inf'), None)

    for move in moves:
        new_pos = (position[0] + move[0], position[1] + move[1]) if maximizing_player \
                  else (opp_position[0] + move[0], opp_position[1] + move[1])

        if 0 <= new_pos[0] < len(world) and 0 <= new_pos[1] < len(world[0]) and world[new_pos[0]][new_pos[1]] != '⬛':
            eval, _ = minimax(new_pos if maximizing_player else position, 
                              opp_position if maximizing_player else new_pos, 
                              goal, depth - 1, not maximizing_player, world, is_good, question)

            if (maximizing_player and eval > best_eval) or (not maximizing_player and eval < best_eval):
                best_eval, best_move = eval, new_pos

    return best_eval, best_move
````

## Problem #1


Implement the game: players, actions, scores, keeping track of turns, when the game is over, how the world changes, and what strategy each player uses (random moves or MINIMAX).


Play the game:
* '🤖' will use MINIMAX with a depth of 2 (simulate I move and then my opponent moves -- that's it) 
* '🤖' utility function = Euclidean distance to '👾' - Euclidean distance to '❤️'
* '👾' will choose actions at random


When the game is over, print out the world (nicer view) with the following changes:
* Replace a '⬜' with '🟥' to indicate the path of '👾', 
* Replace '⬜' with '🟩' to indicate the path of '🤖',
* Places the '🤖' and '👾' icons at their current positions, and
* '❤️' and '⬛' are not replaced.

````python
def run_robot_game(world, good_start, bad_start, goal, moves, bad_strategy, question):
    """
    Runs a turn-based, two-robot game. A bad robot ('👾') tries to zap the good robot ('🤖'),  
    while the good robot aims to reach the goal ('❤️') without being zapped.   
    
    Args:
        world (List[List[str]]): A 2D grid representing the search area where:
            - '⬜' represents an empty cell.
            - '⬛' represents an obstacle (impassable terrain).
            - '❤️' represents the goal.
            - '🤖' represents the good robot.
            - '👾' represents the bad robot.
        good_start (Tuple[int, int]): The starting position (row, col) of the good robot.
        bad_start (Tuple[int, int]): The starting position (row, col) of the bad robot.
        goal (Tuple[int, int]): The goal position (row, col) that the good robot is trying to reach.
        moves (List[Tuple[int, int]]): A list of possible moves, each represented as (row_offset, col_offset).
        bad_strategy (str): The strategy used by the bad robot:
            - 'random': Moves randomly.
            - 'minimax_2': Uses Minimax with depth 2.
            - 'minimax_4': Uses Minimax with depth 4.
        question (Any): Additional parameter that may affect the bad robot’s behavior (used in Minimax).

    Returns:
        None: The function prints the world grid after the game ends. The grid is updated as follows:
            - '🟩' marks the path taken by '🤖'.
            - '🟥' marks the path taken by '👾'.
            - '🤖' and '👾' are placed at their final positions.
            - '❤️' (goal) and '⬛' (obstacles) remain unchanged.
    """
    good_pos, bad_pos = good_start, bad_start
    world_copy = copy.deepcopy(world)

    for _ in range(41):  # Each robot moves up to 41 times
        if good_pos == goal or euclidean_dist(good_pos, bad_pos) <= math.sqrt(2): break 
        _, good_pos = minimax(good_pos, bad_pos, goal, 2, True, world, True, False)  
        world_copy[good_pos[0]][good_pos[1]] = '🟩'

        if bad_strategy.startswith('minimax'): 
            depth = 2 if bad_strategy == 'minimax_2' else 4
            _, bad_pos = minimax(bad_pos, good_pos, goal, depth, True, world, False, question)
        elif bad_strategy == 'random':
            move = random.choice(moves)
            new_pos = (bad_pos[0] + move[0], bad_pos[1] + move[1])
            if 0 <= new_pos[0] < len(world) and 0 <= new_pos[1] < len(world[0]) and world[new_pos[0]][new_pos[1]] != '⬛':
                bad_pos = new_pos
        world_copy[bad_pos[0]][bad_pos[1]] = '🟥'
        
    world_copy[len(world)-1][0], world_copy[len(world)-1][len(world)-1], world_copy[goal[0]][goal[1]] = '🤖', '👾', '❤️'
    print("\n".join("".join(row) for row in world_copy))
````

````python
# Amend arguments (if needed), but this should ONLY print out a nice view of the world when the game is over
run_robot_game(world, good_start, bad_start, goal, moves, "random", 1)
````
